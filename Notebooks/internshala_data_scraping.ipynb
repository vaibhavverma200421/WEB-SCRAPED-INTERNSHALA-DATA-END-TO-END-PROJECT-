```python
# IMPORTING REQUIRED LIBRARIES FOR WEB SRACPING AND DATA CLEANING AND EDA

from httpx import request
import pandas as pd
from bs4 import BeautifulSoup
import lxml
import requests
```

```python
# SPECIFYING LINK

internshala_url = 'https://internshala.com/internships/'
```

```python
# SPECIFYING HEADERS FIRST TO NOT GET BLOCKED BY THE WEB SITE 
# Request Headers (Internshala ko lagega ki request kisi real browser se aa rahi hai, bot se nahi)

HEADER = ({'user_agent' : 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/150.0.0.0 Safari/537.36 Edg/150.0.0.0'})
```

```python
# SENDING REQUEST TO THE WEB SITE      # Internshala URL par HTTP GET request bhej kar page fetch karna

response = requests.get(internshala_url, headers = HEADER)

html_content = response.text   # it will extract the raw html code from the repsonse 
```

```python
# VERIFYING SUCCESS OR FAILURE IN PAGE FETCHING

response
```

```python
# CONVERTING RAW HTML CODE INTO STRUCTURES CODE WITH THE HELP OF BeautifulSoup and lxml for html parsing

soup = BeautifulSoup(html_content, 'lxml')
```

```python
# Displaying the structured HTML content (Soup object)

soup
```

Next Step: `soup.find_all()` se job cards nikalna.

```python
# # Page ke saare internship cards ko extract karne ke liye (class from DevTools)

internship_cards = soup.find_all('div', class_= 'individual_internship')    # individual internship is comman in both class 

print(len(internship_cards))
```

                                                    NEXT STEPS 

job_cards me Loop Chalana: Kal hum for card in job_cards: ka loop lagayenge taaki har ek internship card par baari-baari kaam kar sakein.

Details Extract Karna: Har card ke andar se specific tags/classes dhoondh kar Job Title, Company Name, Location, aur Stipend extract karenge.

Data Structured Karna: Saare extracted data ko ek Dictionary / List of Lists me append (store) karte jayenge.

DataFrame & Export: End me us list ko Pandas DataFrame me badal kar CSV / Excel file me save kar lenge.

```python
'''
First, we will work on only the first card (job_cards[0]) from the entire list (job_cards). Once we learn how to extract data from one card, 
we'll use a for loop to extract data from all cards in a second

'''

first_internship_card = internship_cards[0]

# first_internship_card   # it will print the 1st card html struture

# print(first_internship_card.prettify()[:500])
```

Working on the first internship card first is a safe practice. It makes sure your logic is solid without triggering unwanted errors on the whole page. Once the output from one card is perfect, extracting data from 1,000 cards is just a matter of one loop!

```python
# Extracting title from the 1st internship card


title = first_internship_card.find('h2',class_="job-internship-name")

title = title.text.strip() if title else 'Na'

print(title)
```

```python
# EXTRACTING COMPANY NAME FROM THE FIRST INTERNSHIP CARD

company_name = first_internship_card.find('p', class_='company-name')

company_name = company_name.text.strip() if company_name else 'Na'

print(company_name)
```

```python
# Extracting location from the first internship card 

location_of_internship = first_internship_card.find('div',class_="row-1-item locations")

location_of_internship  = location_of_internship.text.strip() if location_of_internship else 'Na'

print(location_of_internship)
```

```python
# Extracting stipend from the first internship card 

stipend = first_internship_card.find('span',class_="stipend")

stipend = stipend.text.strip() if stipend else 'Na'

print(stipend)
```

```python
# Extracting duration of the first internship card
```

```python

calender_icon = first_internship_card.find('i',class_="ic-16-calendar")   ## Calendar icon (<i class="ic-16-calendar">) ka parent container pakad 
                                                                          #kar usse duration text ('3 Months') extract kar rahe hain
duration = calender_icon.parent.text.strip() if calender_icon else 'Na'

print(duration)
```

```python
# Extracting about_job from first internship card

internship_assignment_icon = first_internship_card.find('div',class_="ic-16-assignment icon")

about = internship_assignment_icon.parent.text.strip() if internship_assignment_icon else 'Na'

print(about)
```

```python
# Extracting skills from the first internship card

internship_skills = first_internship_card.find('div',class_="job_skills")

internship_skills = internship_skills.text.strip() if internship_skills else 'Na'

print(internship_skills)
```

```python
# Extracting posting time from first intership card

post_scedule = first_internship_card.find('i',class_="ic-16-reschedule")

post_time = post_scedule.parent.text.strip() if post_scedule else 'Na'

print(post_time)
```

```python
# ALL THE DETAILS EXTRACTDE FROM FIRST INTERNSHIP CARD

print(f'Internship_title : {title}\n\nCompany_name : {company_name}\n\nLocation : {location_of_internship}\n\nStipend : {stipend}\n\nDuration : {duration}\n\nAbout_internship : {about_internship}\n\nSkills : {internship_skills}\n\nPosted : {post_time}')
```

Now we will Loop Through All Internship Cards
Now that individual extraction logic is verified by extracting all the details from the 1st internship_card, running a loop on `internship_cards` to extract `title`, `company`, `location`, `stipend`, `duration`, and `post_time` into a combined list.

```python

'''Looping through `internship_cards` to scrape title, company, location, stipend, and other job metadata from each card.
Stores all collected information as structured records inside `internship_data`.'''


internship_data = []

for cards in internship_cards:
    
    # Title
    title = cards.find('h2', class_="job-internship-name")
    title = title.text.strip() if title else 'Na'
    
    # Company
    company = cards.find('p',class_="company-name")   
    company = company.text.strip() if company else 'Na'
    
    # Location
    location_icon = cards.find('i',class_="ic-16-map-pin") 
    location = location_icon.parent.text.strip() if location_icon else 'Na'
    
    # Stipend
    stipend = cards.find('span',class_="stipend")
    stipend = stipend.text.strip() if stipend else 'Na'
    
    # Duration
    calender_icon = cards.find('i',class_="ic-16-calendar")
    duration = calender_icon.parent.text.strip() if calender_icon else 'Na'
    
    # About Internship
    about_internship = cards.find('div',class_="text")
    about_internship = about_internship.text.strip() if about_internship else 'Na'
    
    # Required skills
    required_skills = cards.find('div',class_="job_skills")
    required_skills = required_skills.text.strip() if required_skills else 'Na'
    
    # When Posted
    posting_icon = cards.find('i',class_="ic-16-reschedule")
    when_posted = posting_icon.parent.text.strip() if posting_icon else 'Na'
    
    # APPENDING ALL THESE DETAILS IN THE DATA [] and also we are making their list to avoid error - list.append() takes exactly one argument (8 given) 
    
    internship_data.append([title, company, location, stipend, duration, about_internship, required_skills, when_posted])
    
    
```

```python
"""
Prints the complete `internship_data` list to inspect the raw scraped results.
Useful for checking data formatting before converting it into a Pandas DataFrame.
"""


internship_data
```

'''  CONVERTING OUR `internship_data` into a dataframe using Pandas library '''

```python


internshala_data = pd.DataFrame(internship_data,columns=['title', 'company', 'location', 'stipend', 'duration', 'about_internship', 'required_skills', 'when_posted'])

                 #  SAVING TO csv file 

internshala_data.to_csv('internship_data.csv',index= False)
```
